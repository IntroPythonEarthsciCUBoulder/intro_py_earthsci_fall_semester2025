#### Wldfires are increasing with climate change
Wildfires are a yearly occurrence for the Western US, with thousands of fires sparking most commonly during August through November. Dry vegetation and lack of precipitation in these months enhance fire vulnerability. It has been shown that over the past few decades a warming climate lengthens the wildfire season (Dennison et al., 2014). Fire activity not only affects the organisms (inluding humans) that inhabit or use this land, but also impacts the post-fire geomorphological processes and sediment transport.
    


Figure from Denison et al., 2014

#### Wildfires can make landscapes more prone to erosion

The loss of vegetation directly affects the hydrologic cycle of the landscape; with vegetation wiped out, the interception of rainfall will decrease. When fires removes vegetation, it also affects the protection it provides the soil, enabling forces such as wind and rainsplash to affect the surface more intensely. Wildfires also directly make physical and chemical changes to the structure of soil. Fires enhances *soil water repellency* due to ash that affect pores, and due to chemical coatings to grains. <br>





#### How Do Changes Affect the Water Balance Equation?

The water balance in its simplest linear form is written as follows:

$P = Q + ET + \Delta S $

$P =$ precipitation in mm/time

$Q =$ runoff in mm/time

$ET =$ evapotranspiration in mm/time

$\Delta S $ = change in groundwater storage due to infiltration in mm/time

#### Debris Flow 

A debris flow is a moving mass of loose mud, sand, soil, rock, water and air that travels down a slope under the influence of gravity. To be considered a debris flow, the moving material must be loose and capable of "flow," and at least 50% of the material must be sand-size particles or larger.

Because debris flows travel relatively fast and carry a lot of mass, they are very destructive. 
Debris flows are a major hazard in the years directly after a fire. In the Western US, the USGS is mandated to make post-fire risk maps that show the likelihood of debris flow occurrence for diffferent streams.

## What is a List in Python?

So far, we have been assigning single variables. What if you want to have many values, or a dataset with a single name? Lists in python are used to store multiple data items in a single *variable*.<br>

Syntax: Lists are defined by using square brackets, with commas between the elements: 

```
a = [4.0, 2.5, 5.1]
```

To create an empty list, use a pair of square brackets by itself:
```
my_empty_list = []
```

Lists are *mutable*, meaning that they can be changed or updated.

You can add items to a list with `append()`:
```
my_empty_list.append(["I am not empty!"])
print(my_empty_list)

["I am not empty!"]
```

You can combine ("concatenate") lists with the `+` operator:
```
ralphie = [1]
buff = ["yay!"]
print(ralphie + buff)

[1, "yay!"]
```

Lists can contain different datatypes together, so it is a *compound data type*:
```
some_list = ["apples", 3.1415926, 2025, True]
```

You can access items in a list using *indexing*:
```
print(some_list[0])

"apples"

print(some_list[2])

2025

print(type(some_list[3]))

bool
```

You can access a sequence of items in a list using *slicing*:
```
print(some_list[0:2])  # print elements 0 and 1 (NOT 2)

["apples", 3.1415926]

print(some_list[:3])  # everything up to but not including item 3

["apples", 3.1415926, 2025]

print(some_list[2:])  # from item 2 to the end

[2025, True]
```

## Dictionaries

Dictionaries allow you to store key-value pairs. 
Dictionaries are unique in that the keys can not be changed, but the values can be changed. 
To phrase this in python language, the keys arre inmutables, but the values are mutable. 

The syntax: dictionaries use curly brackets, the key is linked to its value by a colon, and key-value pairs are separated by a separated by commas.

Example: CalwoodFire_Streams ={"Geer Canyon": "Lefthand Creek", "Calwood Creek": "South St Vrain" }

So what's different from a list? 

Dictionaries are not indexed by a sequence of numbers, but indexed based on keys. Dictionaries are *associative arrays*.

 

````python
# The streams draining the Calwood Fire go into different drainages
# One side is more used by recreationalist, and drains to Lefthand creek
# the other part goes into the South Saint Vrain, where an water intake of Longmont is located

CalwoodFire_Streams ={"Geer Canyon": "Lefthand Creek", "Central Gulch": "South St Vrain" }
````

````python
# the power of dictionaries is that you can inquire by key instead of index!

#access by key

CalwoodFire_Streams["Central Gulch"]
````

````python
# adding an key-value pair is straightforward

CalwoodFire_Streams ["Heil-Ranch-gully 1"]="Lefthand Creek"
````
