## Miscellaneous coding tips and tricks

### List comprehensions

Python has a neat shorthand called *list comprehensions*. It can be useful for quick filtering. Here's an example:

```
ppt = [0.11, 0.0, 1.34, 2.78, 0.0, 0.0, 0.54]

wet_day_ppt = [x for x in ppt if x > 0.0]
```

You can use it to perform an operation on all elements of a list:

```
vals = [1, 2, 3, 4, 5]
squares = [x**2 for x in vals]
print(squares)
```


Create a sequence of pairs:

```
pairs = [(x, y) for x in range(3) for y in range(3)]
print(pairs)
```

### Dict comprehensions

Imagine you have a list of names, like:
```
epochs = ['Paleocene', 'Eocene', 'Oligocene', 'Miocene', 'Pliocene', 'Pleistocene', 'Holocene']
```
You could look up an epoch by number. But what if you wanted to look up the name associated with the Nth epoch of the Cenozoic era? 
```
epochs_by_number = {k:v for (k, v) in zip(range(len(epochs)), epochs) }
print(epochs_by_number)
```
