# Working with Python dictionaries

Examples of information that comes in **key-value pairs**:

Name | Mass density (kg/m$^3$)
-----|------------------------
Water | 1,000
Quartz | 2,650
Biotite | 3,100

Planet | Radius (km$^2$)
-------|----------------
Mercury | 2,440
Venus | 6,050
Earth | 6,370
Mars | 3,390

Element | Symbol
--------|-------
silicon | Si
oxygen | O
carbon | C
 

```python
radii = {}
type(radii)
```

```python
density = {
    "water" : 1000,
    "quartz" : 2650,
    "biotite" : 3100
}
```

```python
print(density)
```

```python
elements = {
    "oxygen" : "O",
    "silicon" : "Si",
    "carbon" : "C"
}
```

```python
print(elements)
```

```python
density["quartz"]
```

```python
famous_years = {
    1066 : "battle of hastings",
    1776 : "declaration of independence"
}
```
