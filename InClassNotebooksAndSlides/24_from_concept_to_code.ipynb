### Background

#### Mathematical solution for a step-change in temperature

We infer that the curve in the borehole temperature reflects relatively recent warming at the surface. For this exercise, we will imagine that the warming took place in a single abrupt step at some point in the historic past. While that's not necessarily a very realistic assumption---more likely, the ground has warmed progressively over the past several decades---it gives us a convenient starting place.

Here we will take advantage of a mathematical solution to the heat conduction equation that applies to the special case when the temperature at one end of the domain (in this case, the ground surface) undergoes a step change. This equation predicts what the temperature will be at any given depth and at any given time since the step change.

Let $T(z, t)$ be the temperature (in $^\circ$C) at depth $z$ below the ground and time $t$ since the surface warming. $T_0$ will represent the ground-surface temperature at time $t=0$, i.e., just before warming occurred. The symbol $\Gamma$ will be used to represent the geothermal gradient just before surface warming (say 0.04 $^\circ$C/m). The magnitude of surface warming is $\Delta T$ degrees. Finally, $\kappa$ denotes the thermal diffusivity parameter, which for our purposes we can assume to be in the neighborhood of $10^{-6} m^2/s$.

With these symbols defined, the solution for a step change away from a linear geotherm is:

$$
T(z,t) = T_0 + \Gamma z + \Delta T \left[ 1 - \text{erf} \left( \frac{z}{2\sqrt{\kappa t}}\right) \right]
$$

`erf` is a special function called the *error function*, and you can find an (approximate) version in the scipy library. Here's what `erf` looks like:

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import erf

x = np.linspace(-3, 3, 101)
plt.plot(x, erf(x))
```

#### Measuring the misfit between a model and data

One way to measure misfit is with the *root mean square error*:

$$
\text{RMSE} = \sqrt{\frac{1}{n} \sum_{i=1}^n (x_\text{obs} - x_{pred})^2}
$$

In our case, $x_\text{obs}$ represents the temperature measurements at different depths, and $x_\text{pred}$ represents the modeled temperatures at those same depths.
