### Pandas `DataFrame`

Pandas defines a new variable type called a `DataFrame`. A `DataFrame` has some features in common with dictionaries, and some in common with 2D numpy arrays:

- A `DataFrame` contains tabular data in rows and columns
- Each column is identified by a header string
- Columns can be accessed with dictionary-like syntax, using the header string as the key
- Columns can have different data types
- You can add columns the same way you add items to a dictionary
- The Pandas `read_csv()` function reads a CSV file directly into a `DataFrame`
- Pandas data can be plotted using matplotlib functions, or through its own `plot()` function

#### Example 2: a csv file with *two* header lines

We've previously worked with the file `hardangerjokulen2006-2007.csv`, which contains data on summer and winter melt/accumulation of ice on a glacier in Norway, organized by altitude ranges. This file is in **csv** format, which we can handle with Pandas' `read_csv()` function. However, it has *two* header lines.

````python
!cat hardangerjokulen2006-2007.csv
````

````python
import pandas as pd

hardf = pd.read_csv("hardangerjokulen2006-2007.csv")
hardf
````

Because the first row of each column looks like a string, the DataFrame has made all columns strings. How can we get this to be a DataFrame *without* the second header line appearing as data, and with the columns being numbers (`float` data) instead of strings?

We don't really need to keep the info in the second header line, so we can delete that row from the DataFrame. The DataFrame `drop()` method can be used to delete a row or column:

```
hardf = hardf.drop(0)
hardf
```

(Quiz: why do we have to use the assignment operation?)

````python
hardf = hardf.drop(0)
hardf
````

Great! Now all the data in our columns are numbers. But they're still represented as strings, which isn't too convenient. For example, if we tried to calculate the altitude difference by subtracting the max and min altitudes, we'd have issues...

```
hardf["Altitude max"] - hardf["Altitude min"]
```

would produce the error:
```
TypeError: unsupported operand type(s) for -: 'str' and 'str'
```
because Python doesn't have a rule for the `-` operator when the operands are strings (try it!).

````python
hardf["Altitude mid"] = 0.5 * (hardf["Altitude min"] + hardf["Altitude max"])
````

Fortunately, there's a DataFrame method called `astype()` that, by default, converts the data to a given type:

```
hardf = hardf.astype(float)
print(hardf.dtypes)
hardf["Altitude difference"] = hardf["Altitude max"] - hardf["Altitude min"]
hardf
```

That will cast all the rows and columns into `float64`. Now we can treat the columns as numpy arrays to do further analysis. 

````python
hardf = hardf.astype(float)
print(hardf.dtypes)
````

#### Example 3: a tab-delimited file with a long header

We've already played a bit with the Antarctic composite CO2 dataset. Now let's take a closer look at the file. Here's an excerpt:
```
# Antarctic Ice Cores Revised 800KYr CO2 Data
#----------------------------------------------------
#                World Data Service for Paleoclimatology, Boulder
#                                  and
```
(...skipping a bunch of lines here...)
```
# Data line format - tab-delimited text, variable short name as header
# Missing_Values:
age_gas_calBP	co2_ppm	co2_1s_ppm
-51.03	368.02	0.06
-48.00	361.78	0.37
```
This is a *tab-delimited text file*, with a few dozen header lines before the data section begins. All of the header lines are preceded with a `#` character. The first line without a `#` represent the names of the three quantities: `age_gas_calBP` (age of the sample in years before 1950), `co2_ppm` (CO2 concentration in parts per million), and `co2_1s_ppm` (measurement standard deviation ["one sigma" means one standard deviation from the average] in ppm).

To read this into a DataFrame using `read_csv()`, we have to make use of some of that function's optional parameters. Fortunately, there are a lot of these that we can take advantage of! In this case, we need to tell `read_csv()` to ignore lines starting with `#`. For that, we can use the `comment` parameter.

We also need to tell `read_csv()` that this is a *tab*-delimited file, not a *comma*-delimited file. For that, we can use the optional `delimiter` parameter. A tab stop is a non-printing character, but we can use the code `\t` to represent a tab. Here goes:
```
icecoredf = pd.read_csv(
    "antarctica2015co2composite-noaa.txt",
    comment="#",
    delimiter="\t"
)
icecoredf
```

In practice, reading data from a text file into Pandas often takes some monkeying around, and some careful reading of the documentation! If all else fails, you can manually edit the data file to get it into a format that `read_csv()` will handle---but if possible, it's best to leave the data file alone and instead work with the various optional arguments that `read_csv()` provides.

#### Example 4: multiple-space delimited file

The Mauna Loa CO2 data file, called `co2_annmean_mlo_1959_2024.txt`, has column values separated by two or more spaces. We can still read it with `read_csv()` but we have to get into the weeds. Specifically, we need to specify:

1. That "#" means a comment line: `comment="#"`
2. That columns are separated by two or more spaces:
       `sep='[ ]{2,}', engine='python'`
3. The names for the columns (which are hidden behind "#")

After much noodling around, here's one method I found:

````python
mldf = pd.read_csv(
    "co2_annmean_mlo_1959_2024.txt",
    sep='[ ]{2,}',    # this is regexp-speak for "2 or more spaces"
    engine='python',  # not sure why needed...
    comment="#",      # ignore comment lines starting with #
    names=["year", "co2", "uncertainty"]  # give names to columns
)

mldf
````

Lesson: sometimes it takes some study and trial-and-error to get your data in!

## Handling Date and Time stamps

Different programming languages and software packages handle date and time stamps in their own unique ways. Pandas has a set of functions for creating and managing timeseries.

We need to convert the USGS entries in the DateTime column into a format that Pandas can work with. Luckily, the to_datetime function in the Pandas library can convert it directly. Note that this conversion takes up a bit of time, wait till you see those first five lines reported....


````python
# catch up to where we were with this dataset in the last class...
data = pd.read_csv("USGS09058000_discharge_NoHead_20112021.csv")
data.columns = ['Agency', 'SiteNo', 'DateTime', 'Discharge_cfs']
data['Discharge_m3sec'] = data['Discharge_cfs'] * 0.028316847 
data
````

````python
# using the pd.to_datetime functionality
data['NewDateTime'] = pd.to_datetime(data['DateTime'])
data.head()
````

## Creating data subsets and removing columns

We can select specific ranges of our data in both the row and column directions using either label or integer-based indexing.

loc: indexing via labels or integers <br>
iloc: indexing via integers <br>

To select a subset of rows AND columns from our DataFrame, we can use the loc method and the integer indices for both rows and columns. 

````python
data.loc[0:2, ['DateTime', 'Discharge_m3sec']]
````

````python
data.iloc[0:2, -2:]
````

````python
print('With iloc:', data.iloc[0:2, -2:].shape)
print('With loc:', data.loc[0:2, ['DateTime', 'Discharge_m3sec']].shape)
````

Since we can call individual columns (or lists of columns) from a DataFrame, the simplest way to remove columns is by creating a new DataFrame with only the columns we want.

````python
discharge_data = data[['DateTime', 'Discharge_m3sec']]
discharge_data
````
