## Functions
Define the following functions:

A "hello" function which takes in a name, and **prints** a greeting

````python
def hello(name):
    print("greetings", name)

hello("fred")
````

An "add" function which takes in two numbers, and **returns** their sum.

````python
def add(a, b):
    return a + b

add(2, 3)
````

A "radiating temperature" function which takes in a solar irradiance and an albedo and **returns** the blackbody temperature

````python
def radtemp(irrad, albedo):
    sigma = 5.67e-8
    return ((irrad * (1. - albedo)) / (4 * sigma))**0.25

radtemp(1360, 0.31)
````

A "planet info" function which takes in a planet name, a temperature, and a radius and **returns** a string with a sentence detailing that information about the planet.

````python
def planet_info(name, temp, rad):
    print("The planet", name, "has a radius of", rad,
          "km and a temperature of", temp, "deg.")

planet_info("earth", 15, 6370)
````

## if and else

Define a function "grade" which takes in a numerical grade, and returns a letter grade (A, B, C, D, and F)

````python
def grade(score):
    if score >= 90:
        g = "A"
    elif score >= 80:
        g = "B"
    elif score >= 70:
        g = "C"
    elif score >= 60:
        g = "D"
    else:
        g = "F"
    return g

gr = grade(77.3)
print(gr)
````
