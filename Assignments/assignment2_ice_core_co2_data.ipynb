### Part 1A: reading data

*The data we will use are contained in two text-format files:*

- *`antarctica2015co2composite-noaa.txt` contains paleo-CO2 measurements from bubbles in ice cores*
- *`co2_annmean_mlo_1959_2024.txt` contains annual averages of modern measurements of atmospheric CO2 at the Mauna Loa Observatory, Hawaii, for the years 1959 through 2024.*

The code only stores values before 1850. Run the code block to make sure it works. *Once you are confident that the code block works, create a new code cell. In it, create a modified version in which the code below is contained inside a function. The function should take a string containing a file name as a parameter. It should return two lists: the age, and the co2 content. Run your function and make sure you have these two lists.*

````python
# Make filename a parameter in your function
filename = "antarctica2015co2composite-noaa.txt"

# Put these lines of code in your function
with open(filename, "r") as file:
    lines = file.readlines()
age = []
co2 = []
for line in lines:
    if line[0][0] != '#' and line[0][0] != "a":
        split_line = line.split()
        this_age = float(split_line[0])
        if this_age > 150.0: # only keep data before 1800
            age.append(this_age)
            co2.append(float(split_line[1]))

# Add a return statement that returns age and co2
````
